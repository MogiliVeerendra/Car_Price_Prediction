```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
car = pd.read_excel("../../data/interim/transformed_car.xlsx")
car.head()
```

```python
print(car.shape)
print(car.isna().sum())
print(car.duplicated().sum())
```

```python
x = car.drop(columns=["selling_price","name"])
x.head()
```

```python
y = car["selling_price"]
y.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size= 0.2,
    random_state= True
)
x_train.head()
```

```python
print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
x_train.to_excel("../../data/processed/x_train.xlsx",index=False)
```

```python
x_test.to_excel("../../data/processed/x_test.xlsx", index=False)
y_train.to_excel("../../data/processed/y_train.xlsx",index=False)
y_test.to_excel("../../data/processed/y_test.xlsx",index=False)
```
