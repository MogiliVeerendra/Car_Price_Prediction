```python
import pandas as pd
import numpy as np
```

```python
file_path = "../../data/interim/cleaned_car.xlsx"
car = pd.read_excel(file_path)
car
```

```python
car = pd.get_dummies(car, columns=["fuel"], dtype="int")
car.head()
```

```python
car["seller_type"].unique().tolist()
```

```python
car = pd.get_dummies(car,columns=["seller_type"], dtype="int")
car.head()
```

```python
print(car["owner"].unique().tolist())
print(car["owner"].value_counts())
car["owner"] = car["owner"].replace("Test Drive Car", np.nan)
```

```python
car = car.dropna(subset=["owner"])
car["owner"].value_counts()
```

```python
owner_mapping = {
    "First Owner" : 1,
    "Second Owner" : 2,
    "Third Owner" : 3,
    "Fourth & Above Owner": 4
}

car["owner"] = car["owner"].map(owner_mapping)
```

```python
car["owner"].value_counts()
```

```python
car = pd.get_dummies(car, columns=["transmission"], dtype="int")
```
