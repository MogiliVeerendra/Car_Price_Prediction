```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import (Ridge,Lasso,ElasticNet)
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
import joblib
```

```python
x_train = pd.read_excel("../../data/processed/x_train.xlsx")
x_test = pd.read_excel("../../data/processed/x_test.xlsx")
y_train = pd.read_excel("../../data/processed/y_train.xlsx")
x_train.head()
```

```python
y_train.head()
```

```python
# future scaling 
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# linear regression model training
model = LinearRegression()
model.fit(x_train_scaled, y_train)

for features, coeffiecents in zip(x_train.columns,model.coef_[0]):
    print(features, ":", coeffiecents)
print("intercept:",model.intercept_)
joblib.dump(model, "../../models/linear_regression.pkl")
```

```python
# polynomial regression
for degree in range(2,5):
        
    poly = PolynomialFeatures(degree=degree)
    x_train_poly = poly.fit_transform(x_train)
    x_test_poly = poly.transform(x_test)

    poly_model = LinearRegression()
    poly_model.fit(x_train_poly,y_train)

    joblib.dump(poly, f"../../models/polynomial_features_degree{degree}.pkl")

    joblib.dump(poly_model, f"../../models/polynomial_regression_degree{degree}.pkl")
```

```python
# Ridge regression
for alpha in [0.01,0.1,1.0,10,100]:
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(x_train_scaled,y_train)
    joblib.dump(ridge_model, f"../../models/ridge_regression{alpha}.pkl")
```

```python
# lasso regression
for alpha in [1,10,100,1000]:
    lasso_model = Lasso(alpha=alpha,max_iter=100000)
    lasso_model.fit(x_train_scaled,y_train)
    joblib.dump(lasso_model, f"../../models/lasso_regression{alpha}.pkl")
```

```python
# ElasticNet regression
for alpha in [1,10,100,1000]:

    elasticNet_model = ElasticNet(alpha=alpha, l1_ratio=0.5, max_iter=10000)
    elasticNet_model.fit(x_train_scaled,y_train)
    joblib.dump(elasticNet_model, f"../../models/elasticNet_regression_{alpha}.pkl")
```

```python
# Decision Tree Regressor
decision_model = DecisionTreeRegressor(
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf= 10,
    random_state=True
)
decision_model.fit(x_train,y_train)
joblib.dump(decision_model, "../../models/decision_tree_regressor.pkl")
```

```python
# Random Forest Regression
rf_model = RandomForestRegressor(
    max_depth= 10,
    min_samples_split=10,
    min_samples_leaf=10,
    n_estimators= 70,
    random_state= 42
)
rf_model.fit(x_train,y_train)
joblib.dump(rf_model,"../../models/random_forest_regression.pkl")
```
